```python
#Jacob Macy
#Module_3_Lab
#This program creates an object that stores entered values involving details of a car as variables within the object
#The variables year, make, model, doors, and roof are all user-entered values4
class Vehicle():
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type

class Automobile(Vehicle):
    def __init__(self, vehicle_type, year, make, model, doors, roof):
        super().__init__(vehicle_type)
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof
    def __str__(self):
        return '''Vehicle information
    Type: {}
    Year: {}
    Make: {}
    Model: {}
    Doors: {}
    Roof: {}'''.format(self.vehicle_type, self.year, self.make, self.model, self.doors, self.roof)

year = input('Enter vehicle year: ')
make = input('Enter vehicle make: ')
model = input('Enter vehicle model: ')
doors = input('Enter number of doors (2 or 4): ')
roof = input('Enter vehicle roof type: ')
vehicle = Automobile('car', year, make, model, doors, roof)
print(vehicle)
```
